### 4. Launch queries to:
- Check database connection
- Order views
- Add Holiday days to calendar table
- Create a new table "ventas_diarias" with all the merged important info

````python
consulta_proceso="""


/*Order products by family*/
drop table if exists panaderiasalvador.articulos_top;
create table panaderiasalvador.articulos_top as (

	select
	Articulo,
	FAMILIA,
	sum(importe) as importe_total,
	ROW_NUMBER () OVER(partition by FAMILIA order by sum(importe) desc) as orden
	from panaderiasalvador.raw_ventas
	where FechaVenta >= '2021-05-01'
	group by 1,2

);


/*Create a general calendar from 1/1/2017 to 31/12/2023*/
drop table if exists panaderiasalvador.calendario_dias;
create table panaderiasalvador.calendario_dias as (

with recursive cte_calendario as (
	select date('2017-01-01') as calendar_date -- fecha de inicio
	union all
	select date_add(calendar_date, interval 1 day) as calendar_date from cte_calendario
	where date_add(calendar_date, interval 1 day) <= date('2023-12-31') -- condicion de fin, fecha final
)

select
calendar_date as fecha,
year(calendar_date) as fx_anno,
month(calendar_date) as fx_mes,
day(calendar_date) as fx_day,
date_format(calendar_date, '%Y%m') as fx_anno_mes,
date_format(calendar_Date,'%x-%v') as semana -- formato
from cte_calendario

);

/*Add holiday days to the calendar*/
drop table if exists panaderiasalvador.calendario_completo;
create table panaderiasalvador.calendario_completo as (
select
base.*,
festivos.festivo
from panaderiasalvador.calendario_dias base
left join
	(select
	a.*,
	row_number() over(partition by a.fecha order by a.festivo) as orden
	from panaderiasalvador.raw_calendario a
	) festivos
on base.fecha=festivos.fecha
and festivos.orden=1
);

/*Add sellings per day (remove hourly detail)*/

/*Add the product order and if it's in the study date range*/
drop table if exists panaderiasalvador.ventas_diarias;
create table panaderiasalvador.ventas_diarias as (
select
base.familia,
base.tipo,
base.fechaVenta,
calendario.festivo,
base.articulo,
sum(base.precio*base.cantidad)/sum(base.cantidad) as precio, /* Se calcula como el precio promedio del día, ya que hay cambios de precio a lo largo del día*/
articulos.orden as orden_articulo_familia,
case when base.fechaVenta >=date('2021-05-01') then 'S' else 'N' end as in_fecha_estudio,
sum(base.cantidad) as cantidad,
sum(base.importe) as importe

from panaderiasalvador.raw_ventas base

inner join panaderiasalvador.calendario_completo calendario
on base.FechaVenta=calendario.fecha

inner join panaderiasalvador.articulos_top articulos
on base.familia=articulos.familia
and base.articulo=articulos.articulo


group by 1,2,3,4,5,7,8);


/*Create a final view filtering for this study*/
drop view if exists panaderiasalvador.ventas_diarias_estudio_completo;
create view panaderiasalvador.ventas_diarias_estudio_completo as
(select *
from panaderiasalvador.ventas_diarias
where tipo='VENTA'
and in_fecha_estudio='S'
and orden_articulo_familia<=5);


/*Create an extra view with the filtering for the study without the days to predict (May 23rd)*/
drop view if exists panaderiasalvador.ventas_diarias_estudio;
create view panaderiasalvador.ventas_diarias_estudio as
(select *
from panaderiasalvador.ventas_diarias
where tipo='VENTA'
and in_fecha_estudio='S'
and fechaventa < date('2023-05-01')
and orden_articulo_familia<=5);
"""
````

````python
# Select the correct database
cursor = conexion.cursor()
cursor.execute("USE panaderiasalvador")

# Increase the recursive CTE limit for this session
cursor.execute("SET SESSION cte_max_recursion_depth = 3000")

# Separate the SQL queries
consultas_lista = [
    consulta.strip()
    for consulta in consulta_proceso.split(";")
    if consulta.strip()
]

# Execute the queries one by one
for i, consulta in enumerate(consultas_lista, start=1):
    print(f"Executing query {i}/{len(consultas_lista)}...", flush=True)
    cursor.execute(consulta)

# Confirm the changes
conexion.commit()

print("All queries executed successfully!")
````
