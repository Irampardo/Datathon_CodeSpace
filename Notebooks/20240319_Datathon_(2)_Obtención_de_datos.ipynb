```python
import os
from dotenv import load_dotenv

# Load variables from the .env file in the project root
load_dotenv()

USERNAME = os.environ["DB_USERNAME"]
PASSWORD = os.environ["DB_PASSWORD"]
DB_HOST = os.environ["DB_HOST"]
DB_PORT = int(os.environ["DB_PORT"])
DB_NAME = os.environ["DB_NAME"]

LOCATION = (
    float(os.environ["SHOP_LAT"]),
    float(os.environ["SHOP_LON"]),
)
```

### 2. Connect and test queries to database

```python
import pandas as pd
import sqlalchemy as sa
from sqlalchemy import text
from sqlalchemy.engine import URL

class DatabaseConnection:
    def __init__(
        self,
        password,
        dialect="mysql",
        driver="pymysql",
        username=USERNAME,
        host=DB_HOST,
        port=DB_PORT,
        database=DB_NAME
    ):
        connection_url = URL.create(
            drivername=f"{dialect}+{driver}",
            username=username,
            password=password,
            host=host,
            port=port,
            database=database
        )

        self.engine = sa.create_engine(
            connection_url,
            pool_pre_ping=True,
            connect_args={"connect_timeout": 10}
        )
    
    
    def query_to_df(self, query):
        with self.engine.connect() as conn:
            df = pd.read_sql_query(query, conn.connection)
        return df


query = """
SELECT *
FROM panaderiasalvador.ventas_diarias_estudio
WHERE ARTICULO = '3960'
"""

database = DatabaseConnection(PASSWORD)
ventas = database.query_to_df(query)

ventas
```
